```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')


file_path = 'Thursday-WorkingHours-Afternoon-Infilteration.pcap_ISCX.csv' 
data = pd.read_csv(file_path)

print("Dataset Shape:", data.shape)
print("\nDataset Head:")
display(data.head())
print("\nDataset Info:")
display(data.info())
print("\nDataset Statistical Summary:")
display(data.describe())


print("\nMissing Values Per Column:")
display(data.isnull().sum())

print("\nChecking for Infinite Values:")
display((data == np.inf).sum())

print("\nColumn Names:")
print(data.columns.tolist())
```

```python

columns_to_drop = ['Flow ID', 'Source IP', 'Source Port', 'Destination IP', 'Destination Port', 'Timestamp']
data = data.drop(columns=[col for col in columns_to_drop if col in data.columns], errors='ignore')

print(f"Shape after dropping irrelevant columns: {data.shape}")


data = data.dropna()
print(f"Shape after dropping missing values: {data.shape}")

print("\nTotal missing values remaining (should be 0):", data.isnull().sum().sum())

display(data.head())
```

```python
print(data.columns.tolist())
```

```python

data.columns = data.columns.str.strip()

if 'Label' in data.columns:
    data['Label'] = data['Label'].apply(lambda x: 0 if x == 'BENIGN' else 1)
    print("\nLabel encoding done: BENIGN -> 0, Attack -> 1")
else:
    print("\n'Label' column not found!")


import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6,4))
sns.countplot(x='Label', data=data)
plt.title('Class Distribution (0 = Benign, 1 = Attack)')
plt.xlabel('Class')
plt.ylabel('Count')
plt.grid(axis='y')
plt.show()

numeric_data = data.select_dtypes(include=[np.number])

corr = numeric_data.corr()

plt.figure(figsize=(15,10))
sns.heatmap(corr.iloc[:20, :20], cmap='coolwarm', annot=False)
plt.title('Correlation Heatmap (First 20 Features)')
plt.show()
```

```python

from sklearn.preprocessing import StandardScaler
import pandas as pd
import numpy as np

X = data.drop('Label', axis=1)
y = data['Label']

X = X.select_dtypes(include=[np.number])

X = X.replace([np.inf, -np.inf], np.nan)
X = X.dropna()

y = y.loc[X.index]

print("\nCleaned infinities and synced X and y")
print(f"New X shape before scaling: {X.shape}")
print(f"New y shape: {y.shape}")

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("\nFeature scaling completed.")
print(f"Scaled feature shape: {X_scaled.shape}")

X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)
corr_matrix = X_scaled_df.corr().abs()
upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
to_drop = [column for column in upper.columns if any(upper[column] > 0.90)]
print(f"\n⚡ Highly correlated features to drop (correlation > 0.90): {to_drop}")
X_scaled_df.drop(columns=to_drop, axis=1, inplace=True)
print(f"Shape after dropping correlated features: {X_scaled_df.shape}")

display(X_scaled_df.head())
```

```python
data.shape
```

```python

from sklearn.model_selection import train_test_split

# 5.1 Perform stratified split
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled_df,    
    y,              
    test_size=0.30, 
    random_state=42, 
    stratify=y      
)
print(f"Training feature shape: {X_train.shape}")
print(f"Testing feature shape: {X_test.shape}")
print(f"Training label shape: {y_train.shape}")
print(f"Testing label shape: {y_test.shape}")
print("\nTraining set class distribution:")
print(y_train.value_counts(normalize=True))
print("\nTesting set class distribution:")
print(y_test.value_counts(normalize=True))
```

```python

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, roc_auc_score, classification_report, confusion_matrix, roc_curve
import matplotlib.pyplot as plt
import seaborn as sns

rf_model = RandomForestClassifier(random_state=42)
rf_model.fit(X_train, y_train)
y_pred_rf = rf_model.predict(X_test)


if hasattr(rf_model, "predict_proba"):
    proba_rf = rf_model.predict_proba(X_test)
    if proba_rf.shape[1] == 2:
        y_proba_rf = proba_rf[:, 1]
    else:
        y_proba_rf = proba_rf[:, 0]
else:
    y_proba_rf = y_pred_rf

acc_rf = accuracy_score(y_test, y_pred_rf)
roc_rf = roc_auc_score(y_test, y_proba_rf)
print("\n--- Random Forest Performance ---")
print(f"Accuracy: {acc_rf:.4f}")
print(f"ROC-AUC: {roc_rf:.4f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf))

plt.figure(figsize=(5,4))
sns.heatmap(confusion_matrix(y_test, y_pred_rf), annot=True, fmt='d', cmap='Blues')
plt.title("Confusion Matrix - Random Forest")
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()

fpr_rf, tpr_rf, _ = roc_curve(y_test, y_proba_rf)
plt.figure(figsize=(7,5))
plt.plot(fpr_rf, tpr_rf, label='Random Forest')
plt.plot([0,1],[0,1],'k--')
plt.title('ROC Curve - Random Forest')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.legend()
plt.grid()
plt.show()
```

```python

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.optimizers import Adam
from sklearn.metrics import accuracy_score, roc_auc_score, classification_report, confusion_matrix, roc_curve
import matplotlib.pyplot as plt
import seaborn as sns
```

```python


dnn_model = Sequential([
    Dense(128, activation='relu', input_shape=(X_train.shape[1],)),
    Dropout(0.3),
    Dense(64, activation='relu'),
    Dropout(0.2),
    Dense(32, activation='relu'),
    Dense(1, activation='sigmoid')
])

dnn_model.summary()
```

```python

dnn_model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python

history = dnn_model.fit(
    X_train, y_train,
    epochs=20,
    batch_size=128,
    validation_split=0.2,
    verbose=2
)
```

```python

y_pred_dnn = (dnn_model.predict(X_test) > 0.5).astype("int32")
y_proba_dnn = dnn_model.predict(X_test)

acc_dnn = accuracy_score(y_test, y_pred_dnn)
roc_dnn = roc_auc_score(y_test, y_proba_dnn)

print("\n--- DNN Model Performance ---")
print(f"Accuracy: {acc_dnn:.4f}")
print(f"ROC-AUC: {roc_dnn:.4f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred_dnn))

plt.figure(figsize=(5,4))
sns.heatmap(confusion_matrix(y_test, y_pred_dnn), annot=True, fmt='d', cmap='Blues')
plt.title("Confusion Matrix - DNN")
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()

fpr_dnn, tpr_dnn, _ = roc_curve(y_test, y_proba_dnn)
plt.figure(figsize=(7,5))
plt.plot(fpr_dnn, tpr_dnn, label='DNN')
plt.plot([0,1],[0,1],'k--')
plt.title('ROC Curve - DNN')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.legend()
plt.grid()
plt.show()
```
